```python
import torch
import numpy as np
```

```python
for key,value in torch.load('mnist_model.pth').items():
    arr = np.array(value.cpu().numpy().astype(np.float32))
    arr = np.transpose(arr)
    arr.tofile("binary_files/" + key + '.bin')
    
```

```python
target_index = 4  # The 5th element

# 1. Read the 5th label
with open('data/raw/train-labels-idx1-ubyte', 'rb') as f:
    # Seek past the header AND the first 4 labels
    f.seek(8 + target_index)
    label = f.read(1)[0]

# 2. Read the 5th image
with open('data/raw/train-images-idx3-ubyte', 'rb') as f:
    # Seek past the header AND the first 4 images (4 * 784)
    f.seek(16 + (target_index * 784))
    image = np.frombuffer(f.read(784), dtype=np.uint8).astype(np.float32)/255.0
    image.tofile('binary_files/input.bin')
```
